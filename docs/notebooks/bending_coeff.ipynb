The bending mode coefficient is some scalar value that is calculated by multiplying the transpose of the $U$ matrix by whatever surface you are fitting to in FEA space. The resulting value is a scalar coefficient representing how much of some bending mode is present in the surface error.

```python
file_path = Path("C:/Users/solva/OneDrive - University of Arizona/Desktop/bending_mode/stp_tel_166_forcespace_mkII.mat") # define where the forcespace file is on your local machine
force_space = ForceSpace(file_path) # call the forcespace function
bend = Bending(file_path)

psd_tools = psd_utils.PSDUtils() # initialize psd_tools package

modes = 5 # define number of modes you're fitting to

input_coeff = np.arange(1, modes + 1)

input_fea = force_space.data['U'][:, :modes]*input_coeff
#print(np.arange(0, modes))
cols = []

for b in range(modes): # define a for loop to multiply modes to
    rmsForces_bal, b_mode_fit, forces =  bend.fit_bending(modes_to_fit = np.arange(0, modes), map=input_fea)
    #input_fea = force_space.data['U'][:, b] # take some input from the bending modes
    mask = np.isfinite(input_fea)

    print(force_space.data['U'][:, b].T.shape)
    print(input_fea.shape)
    test_coef = np.dot(force_space.data['U'][:, b].T, input_fea)

    new_coef = test_coef.reshape(modes, 1)

    #print(new_coef)
    #print(force_space.data['U'][:, 0].T)
    #print(test_coef)
    cols.append(new_coef)
   

coef_array = np.array(cols)
test_ugh = coef_array.reshape(modes, modes)
print(np.diag(test_ugh))
new_coefs = np.diag(test_ugh)

final = new_coefs/input_coeff
print(final)


#xrange = np.max(force_space.data["x_nodes"]) - np.min(force_space.data["y_nodes"])
#n_nodes = len(force_space.data["x_nodes"])
#dx = xrange / dims[0]
#coords = psd_tools.coord_arrays(dims, dx=dx, offset_x=0, offset_y=0)

#b_mode = force_space.fea_to_grid(input_fea, coords=coords, method="linear")

#b = Bending(file_path)
#modes = 33
#[residual, fitted_surf, fitted_surf_mask, rmsForces, forces, bendCoef] = b.fit_bending(map=b_mode, mask=np.isfinite(b_mode), n_modes = modes, dx = dx, coords = coords, plots=False, method='orig', gauss_remove=False)
#print(bendCoef)
```
